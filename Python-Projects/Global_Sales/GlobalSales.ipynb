**1.Which product categories generate the highest sales revenue?**

```python
df_category_merge['Revenue']=df_category_merge['Unit Price USD']*df_category_merge['Quantity']
df_category_merge
```

**3.Which customer age groups spend the most?**

```python
df_customer
```

```python
df_customer['age'] = (pd.to_datetime('today') - df_customer['Birthday']) // pd.Timedelta(days=365.25)
df_customer
```

```python
def assign_age_group(age):
    if pd.isna(age):
        return 'Unknown'
    elif age <= 17:
        return '0-17'
    elif age <= 29:
        return '18-29'
    elif age <= 44:
        return '30-44'
    elif age <= 59:
        return '45-59'
    else:
        return '60+'
df_customer['age_group'] = df_customer['age'].apply(assign_age_group)
```

```python
age_group_merge=pd.merge(df_customer,df_countries_merge,on='CustomerKey')
age_group_merge
```

```python
Result3=age_group_merge.groupby('age_group')['Revenue'].sum()
Result3
```

```python
Result3.plot(kind='pie')
```
